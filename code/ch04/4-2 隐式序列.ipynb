# 隐式序列

- 序列（sequence）的表示，不必把每个元素都显式地存进计算机内存。
- 也就是说，我们可以构造一个对象，它提供对某个序列数据集里所有元素的访问，却不必提前算出每个元素的值；元素在需要时才计算。

## 可迭代对象

任何能产生迭代器的值都称为可迭代值（iterable value）。在 Python 中，可迭代值是任何能传给内置 iter 函数的值。
可迭代对象包括:
- **序列值**:例如字符串(string)和元组(tupe)
- **容器**:例如集合(set)和字典(dict)

```python
d = {'one': 1, 'two': 2, 'three': 3}
d
```

```python
k = iter(d) # k是字典d的迭代器
```

```python
next(k)
```

```python
v = iter(d.values()) # v是字典d的值的迭代器
```

```python
next(v)
```

```python
d.pop('two')
# next(k) # dictionary changed size during iteration
```

## 内置迭代器

```python
# map、filter、zip、enumerate等接受可迭代值作为参数,返回可迭代对象
def double_and_print(x):
    print('***', x, '=>', 2*x, '***')
    return 2*x

s = range(3, 7)
doubled = map(double_and_print, s) # doubled是一个可迭代对象
```

```python
next(doubled)
```

```python
next(doubled)
```

```python
list(doubled) # 迭代器只能被迭代一次
```

## For语句

```python
# for <name> in <expression>:
#     <suite>
counts = [1, 2, 3]
for count in counts:
    print(count)
```

```python
# for执行过程
iterator = counts.__iter__()
try:
    while True:
        count = iterator.__next__()
        print(count)
except StopIteration:
    pass
```

## 生成器和Yield语句

```python
def letters_generator():
    start = 'a'
    while start <= 'z':
        yield start
        start = chr(ord(start) + 1)
```

```python
for letter in letters_generator():
    print(letter)
```

```python
letters = letters_generator()
type(letters)
```

```python
letters.__next__()
```

## 可迭代接口

```python
class LetterIter:
    def __init__(self, start='a', end='z'):
        self.next_letter = start
        self.end = end
    
    def __next__(self):
        if self.next_letter == self.end:
            raise StopIteration
        letter = self.next_letter
        self.next_letter = chr(ord(self.next_letter) + 1)
        return letter


class Letters:
    def __init__(self, start='a', end='z'):
        self.start = start
        self.end = end
    
    def __iter__(self):
        return LetterIter(self.start, self.end)
```

```python
b_to_k = Letters('b', 'k')
```

```python
first_iter = b_to_k.__iter__()
next(first_iter)
```

```python
next(first_iter)
```

```python
second_iter = b_to_k.__iter__()
next(second_iter)
```

```python
caps = map(lambda x: x.upper(), b_to_k)
```

```python
next(caps)
```

## 用yield创建可迭代对象

```python
# 迭代器只进行一次遍历
# 部分情况需要多次遍历
def all_pairs(s):
    for x in s:
        for y in s:
            yield (x, y)
```

```python
list(all_pairs([1, 2, 3]))
```

```python
class LettersWithYield:
    def __init__(self, start='a', end='z'):
        self.start = start
        self.end = end

    def __iter__(self):
        next_letter = self.start
        while next_letter <= self.end:
            yield next_letter
            next_letter = chr(ord(next_letter) + 1)
```

```python
letters = LettersWithYield()
list(all_pairs(letters))[:5]
```

## Python流

```python
class Stream:
    """惰性链接"""
    class empty:
        def __repr__(self):
            return "Stream.empty"
    empty = empty()
    def __init__(self, first, compute_rest: lambda: empty):
        assert callable(compute_rest), "compute_rest 必须可调用"
        self.first = first
        self._compute_rest = compute_rest
    @property
    def rest(self):
        """返回Stream的其它部分,如果需要则计算"""
        if self._compute_rest is not None:
            self._rest = self._compute_rest()
            self._compute_rest = None
        return self._rest
    def __repr__(self):
        return 'Stream({0}, <...>)'.format(repr(self.first))
```

```python
s = Stream(1, lambda: Stream(2 + 3, lambda: Stream(9)))
```

```python
s.first
s.rest.first
```

```python
def integer_stream(first):
    def compute_rest():
        return integer_stream(first + 1)
    return Stream(first, compute_rest)
```

```python
positives = integer_stream(1)
positives
```

```python
positives.first
```

```python
positives.rest.first
positives.rest.rest
```

```python
def map_stream(fn, s):
    if s is Stream.empty:
        return s
    def compute_rest():
        return map_stream(fn, s.rest)
    return Stream(fn(s.first), compute_rest)
```

```python
def filter_stream(fn, s):
    if s is Stream.empty:
        return s
    def compute_rest():
        return filter_stream(fn, s.rest)
    if fn(s.first):
        return Stream(s.first, compute_rest)
    else:
        return compute_rest()
```

```python
def first_k_as_list(s, k):
    first_k = []
    while s.first != Stream.empty and k > 0:
        first_k.append(s.first)
        s, k = s.rest, k - 1
    return first_k
```

```python
s = integer_stream(3)
s
```

```python
m = map_stream(lambda x: x * x, s)
m
```

```python
first_k_as_list(m, 5)
```
