## 解析表达式

### 词法分析器

```python
from scalc.buffer import Buffer
from scalc.scheme_tokens import tokenize_line, tokenize_lines

# 把字符串解释成词元序列叫做词法分析器
tokenize_line('(+ 1 (* 2.3 45))')
```

### 语法分析器

```python
lines = ['(+ 1', '   (* 2.3 45))']
expression = scheme_read(Buffer(tokenize_lines(lines)))
expression
```

## 计算器求值

```python
from scalc.scalc import calc_apply, calc_eval, as_scheme_list, read_eval_print_loop
```

```python
calc_apply('+', as_scheme_list(1, 2, 3))
calc_apply('-', as_scheme_list(10, 1, 2, 3))
calc_apply('*', nil)
calc_apply('*', as_scheme_list(1, 2, 3, 4, 5))
calc_apply('/', as_scheme_list(40, 5))
```
