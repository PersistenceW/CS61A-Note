### `1.对象隐喻`

对象（Objects）将数据值与行为结合在一起。
- 对象不仅代表信息，而且行为表现也像它们所代表的事物。
- 对象如何与其他对象交互的逻辑，与编码对象值的信息一同被打包在一起。
- 当打印一个对象时，它知道如何以文本形式拼写出自己。
- 如果一个对象由多个部分组成，它知道如何按需展示这些部分。
- 对象既是信息也是过程，它们捆绑在一起，以代表复杂事物的属性、交互和行为。

### `4.局部状态`

```python
def make_withdraw(balance):
    """
        访问 nonlocal 声明的变量名称并不需要使用非局部语句
        相比之下，只有在非局部语句之后，函数才能更改这些帧中名称的绑定。
    """
    def withdraw(amount):
        nonlocal balance
        if amount > balance:
            return 余额不足
        balance = balance - amount
        return balance
    return withdraw
```

```python
withdraw = make_withdraw(100)
withdraw(25)
```

```python
withdraw(25)
```

- 非局部 Non-local 赋值对我们意识到程序是由独立、自治的对象组成的至关重要，这些对象相互交互但又各自维护自己的内部状态

### `7.列表和字典的实现`

```python
empty = ()

def link(first_value, rest_link):
    """ 创建一个链表节点 """
    return [first_value, rest_link]

def first(s):
    """ 返回链表的第一个节点 """
    if s == empty:
        raise IndexError("不能获取空链表的第一个元素")
    return s[0]

def rest(s):
    """ 返回除链表第一个节点外的剩余节点 """
    if s == empty:
        raise IndexError("空链表没有剩余部分")
    return s[1]

def len_link(s):
    """ 返回链表的长度 """
    length = 0
    while s != empty:
        length += 1
        s = rest(s)
    
    return length

def getitem_link(s, index):
    """ 返回链表中指定下标的元素 """
    if not isinstance(index, int):
        raise TypeError("链表下标必须是整数")
    
    if index < 0:
        raise IndexError("下标不能为负数")
    
    while s != empty:
        if index == 0:
            return first(s)
        
        s = rest(s)
        index -= 1
    
    raise IndexError("链表下标越界")

def join_link(s, separator):
    """ 将链表的内容连接成字符串 """
    elements = []

    while s != empty:
        elements.append(str(first(s)))

        s = rest(s)
    return separator.join(elements)
```

```python
def mutable_link():
    """返回一个可变链表的消息分派函数"""
    contents = empty

    def dispatch(message, value=None):
        nonlocal contents

        if message == "len":
            return len_link(contents)

        elif message == "getitem":
            return getitem_link(contents, value)

        elif message == "push_first":
            contents = link(value, contents)

        elif message == "pop_first":
            f = first(contents)
            contents = rest(contents)
            return f

        elif message == "str":
            return join_link(contents, ", ")

        else:
            raise ValueError(f"未知消息：{message}")

    return dispatch
```

```python
s = mutable_link()

s("push_first", 10)
s("push_first", 20)
s("push_first", 30)

print(s("str"))         # 30, 20, 10
print(s("len"))         # 3
print(s("getitem", 0))  # 30
print(s("getitem", 1))  # 20

print(s("pop_first"))   # 30
print(s("str"))         # 20, 10
print(s("len"))         # 2
```

```python
# 字典实现
def dictionary():
        """返回一个字典的函数实现"""
        records = []
        def getitem(key):
            matches = [r for r in records if r[0] == key]
            if len(matches) == 1:
                key, value = matches[0]
                return value
        def setitem(key, value):
            nonlocal records
            non_matches = [r for r in records if r[0] != key]
            records = non_matches + [[key, value]]
        def dispatch(message, key=None, value=None):
            if message == 'getitem':
                return getitem(key)
            elif message == 'setitem':
                setitem(key, value)
        return dispatch
```

```python
d = dictionary()
d('setitem', 3, 9)
d('setitem', 4, 16)
d('getitem', 3)
d('getitem', 4)
```

### `8.调度字典` 

```python
def account(init_balance):
    def withdraw(amount):
        if amount > dispatch['balance']:
            return '余额不足'
        dispatch['balance'] = dispatch['balance'] - amount
        return dispatch['balance']
    def deposit(amount):
        dispatch['balance'] = init_balance + amount
        return init_balance
    dispatch = {
        "balance": init_balance,
        "withdraw": withdraw,
        "deposit": deposit
    }
    return dispatch
```

```python
account = account(100)
account['balance']
```

```python
account['withdraw'](25)
```

```python
account['withdraw'](25)
```

```python
account['deposit'](50)
```
