### ```序列```

序列(sequence)是有序的值的集合。它们拥有共同的行为
- 长度（Length）：序列的长度是有限的，空序列的长度为 0
- 元素选择（Element selection）：序列拥有与每一个小于其长度的非负整数索引相对应的元素，索引从 0 开始表示第一个元素

### `2.序列遍历`

````python
def count(s, value):
    total, index = 0, 0
    while index < len(s):
        if s[index] == value:
            total += 1
        index += 1
    return total

count(digits, 8)
````

````python
# 使用for语句直接遍历元素
def count(s, value):
    total = 0
    for elem in s:
        if elem == value:
            total += 1
    return total

count(digits, 8)
````

`for`语句形式:
```python
for <name> in <expression>:
    <suite>
```
求值过程:
1. 求值头部 (header) 的 `<expression>`，它必须产生一个可迭代 (iterable) 值
2. 对该可迭代值中的每个元素，按顺序执行：
- 在当前帧中将 `<name>` 绑定到该值
- 执行语句组 `<suite>`

````python
# 序列解包
pairs = [[1, 2], [2, 2], [2, 3], [4, 4]]
same_count = 0
for x, y in pairs:
    if x == y:
        same_count += 1
same_count
````

### `3.序列处理`

````python
# 列表推导式
odds = [1, 3, 5, 7, 9]
[x + 1 for x in odds]
````

````python
# 列表推导式的一般形式
# [<映射表达式> for <名称> in <序列表达式> if <过滤表达式>] 
````

````python
# 完全数
# 求因子
def divisors(n):
    return [1] + [x for x in range(2, n) if n % x == 0]

divisors(12)
````

````python
# 求1-1000之间的完全数
[n for n in range(1, 1000) if n == sum(divisors(n)) ]
````

````python
# 给定面积求矩形最小周长
def width(area, height):
    assert area % height == 0
    return area // height

def perimeter(width, height):
    return 2 * (width + height)

def minimum_perimeter(area):
    heights = divisors(area)
    perimeters = [perimeter(width(area, height), height) for height in heights]
    return min(perimeters)

area = 80
width(area, 5)

perimeter(16, 5)

minimum_perimeter(area)
````

````python
from operator import mul
# 高阶函数
# 映射
def apply_to_all(map_fn, s):
    return [map_fn(x) for x in s]

# 过滤
def keep_if(filter_fn, s):
    return [x for x in s if filter_fn(x)]

def reduce(reduce_fn, s, initial):
    reduced = initial
    for x in s:
        reduced = reduce_fn(reduced, x)
    return reduced

reduce(mul, [2, 4, 8], 1)
````

````python
from operator import add
# 使用高阶函数来寻找完全数
def divisor_of(n):
    divides_n = lambda x: n % x == 0
    return [1] + keep_if(divides_n, range(2, n))

def sum_of_divisor(n):
    return reduce(add, divisor_of(n), 0)

def perfect(n):
    return sum_of_divisor(n) == n

keep_if(perfect, range(1, 1000))
````

````python
# apply_to_all -> map
apply_to_all = lambda map_fn, s : list(map(map_fn, s))
# keep_if -> filter
keep_if = lambda filter_fn, s : list(filter(filter_fn, s))
````

### `6.树`

````python
# 树必须有一个根标签和一个分支
def tree(root_label, branches=[]):
    for branch in branches:
        assert is_tree(branch), '分支必须是树'
    return [root_label] + list(branches)

def label(tree):
    return tree[0]

def branches(tree):
    return tree[1:]
````

````python
def is_tree(tree):
    if type(tree) != list or len(tree) < 0:
        return False
    for branch in branches(tree):
        if not is_tree(branch):
            return False
    return True
````

````python
def is_leaf(tree):
    return not branches(tree)
````
