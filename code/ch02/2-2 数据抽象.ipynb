### `1.有理数`

<分子>/<分母>
让我们先假设已经有了从分子和分母构造有理数的方法，以及给定一个有理数提取其分子和分母的方法。我们要进一步假设，这个构造函数和选择函数通过以下三个函数提供：
- `rational(n, d)` 返回分子为 `n`、分母为 `d` 的有理数
- `numer(x)` 返回有理数 `x` 的分子
- `denom(x)` 返回有理数 `x` 的分母

```python
def add_rationals(x, y):
    nx, dx = numer(x), denom(x)
    ny, dy = numer(y), denom(y)
    return rational(nx * dy + dx * ny, dx * dy) 

def mul_rationals(x, y):
    return rational(numer(x) * numer(y), denom(x) * denom(y))

def print_rational(x):
    print(numer(x), '/', denom(x))

def rationals_are_equal(x, y):
    return numer(x) * denom(y) == denom(x) * numer(y)
```

### `2.对`

```python
from math import gcd
def rational(n, d):
    g = gcd(n, d)
    return [n // g, d // g]

def numer(x):
    return x[0]

def denom(x):
    return x[1]
```

```python
half = rational(1, 2)
print_rational(half)
```

```python
third = rational(1, 3)
print_rational(mul_rationals(half, third))
print_rational(add_rationals(third, third))
```

### `3.抽象屏障`

```python
def square_rational(x):
    return mul_rationals(x, x)
```

### `4.数据的属性`

```python
def pair(x, y):
    def get(index):
        if index == 0:
            return x
        if index == 1:
            return y
    return get

def select(p, i):
    return p(i)
```

```python
p = pair(10, 20)
select(p, 0)
```
