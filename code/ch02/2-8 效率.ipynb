```python
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)
```

```python
fib(5)
```

```python
def count(f):
    def counted(n):
        counted.call_cnt += 1
        return f(n)
    counted.call_cnt = 0
    return counted
```

```python
fib = count(fib)
fib(19)
```
