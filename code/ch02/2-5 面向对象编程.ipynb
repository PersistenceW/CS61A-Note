```python
"""
    class <name>:
        <suite>
"""
```

```python
class Account:
    """A bank account class with a balance."""

    def __init__(self, account_holder):
        self.balance = 0
        self.holder = account_holder
```

```python
# 类实例化以后,使用.来访问属性和方法
a = Account("kirk")
a.balance, a.holder
```

```python
# 每一个实例都有自己的身份标识
b = Account("spock")
b.balance = 200
[acc.balance for acc in (a, b)]
```

```python
# is 判断地址
# == 判断内容
a is a
# a is not b
```

```python
# 绑定到新的变量,不会创建新的对象
c = a
c is a
```

```python
class Account:
    """A bank account class with a balance."""

    def __init__(self, account_holder):
        self.balance = 0
        self.holder = account_holder

    def deposit(self, amount):
        self.balance = self.balance + amount
        return self.balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            return 'Insufficient funds'
        self.balance = self.balance - amount
        return self.balance
```

```python
spock_account = Account('Spock')
spock_account.deposit(1000)
```

```python
spock_account.withdraw(500)
```

```python
spock_account.withdraw(600)
```

### `3.消息传递和点表达式`

在类中定义的方法和在构造函数中分配的实例属性是面向对象编程的基本元素。
- 消息传递的主要思想是，数据值应该通过响应与其表示的抽象类型相关的消息来具有行为。
- 点表达式是 Python 的一个语法特征，它形式化了消息传递隐喻。

```python
# 函数getattr(obj, name)返回对象obj的属性name的值,如果没有该属性则抛出异常
getattr(spock_account, 'balance')
```

```python
# hasattr(obj, name)返回对象obj是否有属性name
hasattr(spock_account, 'deposit')
```

函数和方法
- 在对象上调用方法时，该对象将作为第一个参数隐式传递给该方法
- Python 区分了函数（我们从文本开头就一直在创建）和绑定方法（将函数和将调用该方法的对象耦合在一起）

```python
print(type(Account.deposit)) # <class 'function'>
print(type(spock_account.deposit)) # <class 'method'>
```

```python
Account.deposit(spock_account, 1000)
```

```python
spock_account.deposit(1000)
```

### `5.继承`
- 子类继承其父类的属性，但可以重写某些属性，包括某些方法
- 对于继承，我们只指定子类和父类之间的区别。我们在子类中未指定的任何内容都会被自动假定为与父类的行为一样。

### `6.使用多继承`

```python
class Account:
    """A bank account class with a balance."""

    interest = 0.02  # A class attribute

    def __init__(self, account_holder):
        self.balance = 0
        self.holder = account_holder

    def deposit(self, amount):
        self.balance = self.balance + amount
        return self.balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            return 'Insufficient funds'
        self.balance = self.balance - amount
        return self.balance
```

```python
class CheckingAccount(Account):
    """A bank account that charges for withdrawals."""
    
    withdraw_fee = 1
    interest = 0.01  # Overriding the class attribute

    def withdraw(self, amount):
        return Account.withdraw(self, amount + self.withdraw_fee)
```

```python
checking = CheckingAccount('Jack')
checking.deposit(20)
```

```python
checking.withdraw(5)
```

```python
checking.interest
```

### `7.多继承`

```python
class SavingAccount(Account):
    deposit_charge = 2
    def deposit(self, amount):
        return Account.deposit(self, amount - self.deposit_charge)
```

```python
class AsSeenOnTVAccount(CheckingAccount, SavingAccount):
    def __init__(self, account_holder):
        self.holder = account_holder
        self.balance = 1
```

```python
such_a_deal = AsSeenOnTVAccount('David')
such_a_deal.deposit(20)
```

```python
such_a_deal.withdraw(5)
```

```python
such_a_deal.withdraw_fee
```

```python
such_a_deal.deposit_charge
```

- 换句话说，Python 的对象系统提供了一种方便而灵活的方法来创建和操作对象，使程序员能够更好地组织和管理复杂的程序。
