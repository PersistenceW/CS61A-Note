```python
def pressure(v, t, n):
    """
    计算理想气体的压力（单位：帕斯卡）

    使用理想气体定律：http://en.wikipedia.org/wiki/Ideal_gas_law

    v -- 气体体积，单位：立方米
    t -- 绝对温度，单位：开尔文
    n -- 气体粒子数
    """
    k = 1.38e-23  # 玻尔兹曼常数
    return n * k * t / v
```

```python
# 默认参数
def pressure(v, t, n=6.022e23):
    """
    计算理想气体的压力（单位：帕斯卡）

    使用理想气体定律：http://en.wikipedia.org/wiki/Ideal_gas_law

    v -- 气体体积，单位：立方米
    t -- 绝对温度，单位：开尔文
    n -- 气体粒子数，默认为1
    """
    k = 1.38e-23  # 玻尔兹曼常数
    return n * k * t / v
```

```python
pressure(1, 300)  # 使用默认参数 n=6.022e23
```
