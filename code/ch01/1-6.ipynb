### `1.函数作为参数`

```python
# 求和
def sum_naturals(n):
    total, k = 0, 1
    while k <= n:
        total, k = total + k, k + 1
    return total

# 求立方和
def sum_cubes(n):
    total, k = 0, 1
    while k <= n:
        total, k = total + k * k *k, k + 1
    return total

# pi
def pi_sum(n):
    total, k = 0, 1
    while k <= n:
        total, k = total + 8 / (k * (k + 2) * (k + 4)), k + 4
    return total
```

```python
'''
函数抽象
def <name>(n):
    total, k = 0, 1
    while k <= n:
        total, k = total + <term>(k), k + 1
    return total
'''
def summation(n, term):
    total, k = 0, 1
    while k <= n:
        total, k = total + term(k), k + 1
    return total

def cube(k):
    return k * k * k

def sum_cubes(n):
    return summation(n, cube)

sum_cubes(3)
```

```python
def identity(k):
    return k

def sum_naturals(n):
    return summation(n, identity)

sum_naturals(100)
```

```python
def pi_term(k):
    return 8 / (k * (k + 2) * (k + 4))

def pi_sum(n):
    return summation(n, pi_term)

pi_sum(100)
```

### `2.函数作为通用方法`

```python
# 迭代改进
def improve(update, close, guess=1):
    while not close(guess):
        guess = update(guess)
    return guess
```

```python
# 求解黄金分割率
def golden_update(guess):
    return 1 / guess + 1

def square_close_to_successor(guess):
    return approx_eq(guess * guess, guess + 1)

def approx_eq(x, y, tolerance=1e-15):
    return abs(x - y) < tolerance

improve(golden_update, square_close_to_successor)
```

```python
from math import sqrt
phi = 1 / 2 + sqrt(5) / 2
def improve_test():
    approx_phi = improve(golden_update, square_close_to_successor)
    assert approx_eq(approx_phi, phi), 'Test failed: expected {}, got {}'.format(phi, approx_phi)
improve_test()
```

### `3.定义函数III 嵌套定义`

```python
def average(x, y):
    return (x + y) / 2

def sqrt_update(x, a):
    return average(x, a / x)
```

```python
def sqrt(a):
    def sort_update(x):
        return sqrt_update(x, a)
    def sort_close(x):
        return approx_eq(x * x, a)
    return improve(sort_update, sort_close)
```

```python
sqrt(4)
```

### `4.函数作为返回值`

```python
def compose1(f, g):
    def h(x):
        return f(g(x))
    return h
```

```python
def square(x):
    return x * x

def successor(x):
    return x + 1

square_of_successor = compose1(square, successor)
square_of_successor(5)
```

### `5.示例：牛顿法`

```python
def newton_update(f, df):
    def update(x):
        return x - f(x) / df(x)
    return update
```

```python
def find_zero(f, df):
    def near_zero(x):
        return approx_eq(f(x), 0)
    return improve(newton_update(f, df), near_zero)
```

```python
def square_root_newton(a):
    def f(x):
        return x * x - a
    def df(x):
        return 2 * x
    return find_zero(f, df)

square_root_newton(64)
```

```python
def power(x, n):
    product, k = 1, 0
    while k < n:
        product, k = product * x, k + 1
    return product

def nth_root_of_a(n, a):
    def f(x):
        return power(x, n) - a
    def df(x):
        return n * power(x, n - 1)
    return find_zero(f, df)
```

```python
nth_root_of_a(3, 27)
nth_root_of_a(6, 64)
```

### `6.柯里化`

```python
# 柯里化f(x, y) => g(x)(y)
def curried_pow(x):
    def h(y):
        return power(x, y)
    return h
```

```python
curried_pow(2)(3)  # 8
```

```python
def map_to_range(start, end, f):
    while start < end:
        print(f(start))
        start += 1
```

```python
map_to_range(0, 5, curried_pow(2))  # 1, 2, 4, 8, 16
```

### `9.装饰器`

```python
def trace(fn):
    def wrappered(x):
        print('->', fn, '(', x, ')')
        return fn(x)
    return wrappered

@trace
def triple(x):
    return 3 * x

triple(12)
```
