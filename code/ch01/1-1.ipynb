```python
from urllib.request import urlopen
```

```python
shakespear = urlopen("http://composingprograms.com/shakespeare.txt")
```

```python
words = set(shakespear.read().decode().split())
words
```

```python
test = (w for w in words if len(w) == 6 and w[::-1] in words)
for t in test:
    print(t)
```
